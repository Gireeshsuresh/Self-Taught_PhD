# 🧠 I. Theoretical Foundations

## 🧮 1. Computational Complexity

Computational complexity theory provides the mathematical foundation for analyzing algorithm efficiency and resource requirements. This section covers advanced complexity analysis techniques essential for expert-level algorithm design and optimization.

### ⏱️ Time Complexity Analysis

Time complexity measures how the running time of an algorithm increases with input size. For expert practitioners, understanding both theoretical bounds and practical performance characteristics is crucial.

#### ∞ Asymptotic Notation

- **Big-O Notation ($O$)**: Upper bound on growth rate
  - **Formal Definition**: $f(n) = O(g(n))$ if $\exists c > 0, n_0 > 0$ such that $f(n) \leq c \cdot g(n)$ for all $n \geq n_0$
  - **Practical Interpretation**: Worst-case scenario performance guarantee
  - **Common Misconception**: $O(n^2)$ doesn't mean "exactly $n^2$ operations" but "at most proportional to $n^2$"

- **Omega Notation ($\Omega$)**: Lower bound on growth rate
  - **Formal Definition**: $f(n) = \Omega(g(n))$ if $\exists c > 0, n_0 > 0$ such that $f(n) \geq c \cdot g(n)$ for all $n \geq n_0$
  - **Practical Interpretation**: Best-case scenario performance requirement
  - **Use Case**: Proving that no algorithm can do better than $\Omega(n \log n)$ for comparison-based sorting

- **Theta Notation ($\Theta$)**: Tight bound
  - **Formal Definition**: $f(n) = \Theta(g(n))$ if $f(n) = O(g(n))$ and $f(n) = \Omega(g(n))$
  - **Practical Interpretation**: Both upper and lower bounds match, providing precise characterization
  - **Example**: Merge sort is $\Theta(n \log n)$ - this is both the best and worst case

- **Little-o Notation ($o$)**: Strictly smaller than
  - **Formal Definition**: $f(n) = o(g(n))$ if $\lim_{n \to \infty} f(n)/g(n) = 0$
  - **Practical Interpretation**: $f$ grows strictly slower than $g$
  - **Example**: $n = o(n \log n)$, but $n \neq o(n)$

#### 📊 Advanced Complexity Classes

| Complexity    | Notation   | Description                | Examples                        | Practical Implications         |
|--------------|------------|----------------------------|----------------------------------|-------------------------------|
| Constant     | $O(1)$     | Independent of input size  | Array access, hash table lookup  | Ideal for frequent operations |
| Logarithmic  | $O(\log n)$| Grows logarithmically      | Binary search, balanced trees    | Excellent scalability         |
| Linear       | $O(n)$     | Grows proportionally       | Linear search, single pass       | Acceptable for most apps      |
| Linearithmic | $O(n \log n)$| Linear × logarithmic     | Merge sort, heap sort           | Optimal for many problems     |
| Quadratic    | $O(n^2)$   | Grows quadratically        | Bubble sort, nested loops        | Problematic for large inputs  |
| Cubic        | $O(n^3)$   | Grows cubically            | Matrix multiplication (naive)    | Usually needs optimization    |
| Exponential  | $O(2^n)$   | Grows exponentially        | Brute force subset generation    | Intractable for large inputs  |
| Factorial    | $O(n!)$    | Grows factorially          | Traveling salesman (brute force) | Completely intractable        |

#### 🏁 Complexity Hierarchy and Practical Implications

**P vs NP Problem:** The fundamental question of whether problems solvable in polynomial time (P) equal those verifiable in polynomial time (NP).

**Practical Complexity Guidelines:**
- $O(1)$ to $O(\log n)$: Excellent for any input size
- $O(n)$ to $O(n \log n)$: Good for inputs up to $10^6$-$10^7$ elements
- $O(n^2)$: Acceptable for inputs up to $10^3$-$10^4$ elements
- $O(n^3)$: Marginal for inputs up to $10^2$-$10^3$ elements
- $O(2^n)$ and above: Only for very small inputs ($n < 30$)

### 💾 Space Complexity Analysis

Space complexity measures the amount of memory an algorithm requires relative to input size. Expert practitioners must consider both theoretical space bounds and practical memory constraints.

**Space Complexity Categories:**
- **Auxiliary Space:** Extra space used by the algorithm (excluding input space)
- **Input Space:** Space required to store the input (often not counted in analysis)
- **Recursion Stack Space:** Space used by recursive calls
- **Working Space:** Temporary variables and intermediate results

**Space-Time Tradeoffs:**
- **Space for Time:** Use extra memory to reduce time complexity
- **Time for Space:** Use more time to reduce space requirements
- **Example:** Hash table uses $O(n)$ space for $O(1)$ average lookup time

### ⚖️ Amortized Analysis

Amortized analysis provides average-case performance guarantees for data structures with varying operation costs. This is crucial for understanding the true performance of dynamic data structures.

**Methods of Amortized Analysis:**

1. **Aggregate Method:**
   - Calculate total cost of $n$ operations
   - Amortized cost = total cost $\div n$
   - **Example:** Dynamic array resizing—some operations are expensive, but average is $O(1)$

2. **Accounting Method:**
   - Assign charges to operations
   - Ensure total charges $\geq$ actual costs
   - **Example:** Stack with multipop—charge extra for push to pay for multipop

3. **Potential Method:**
   - Define potential function $\Phi(D)$ for data structure state $D$
   - Amortized cost = actual cost $+ \Phi(D') - \Phi(D)$
   - **Example:** Fibonacci heap—potential function based on tree structure

**Real-World Applications:**
- **Dynamic Arrays:** Amortized $O(1)$ append operation
- **Hash Tables:** Amortized $O(1)$ operations with resizing
- **Disjoint Set Union:** Amortized $O(\alpha(n))$ per operation ($\alpha$ is inverse Ackermann)

## 📐 2. Mathematical Foundations

Mathematical concepts that underpin sophisticated algorithmic techniques. Expert practitioners must master these mathematical tools to design and analyze complex algorithms effectively.

### 🧩 Proof Techniques

- **Mathematical Induction:** The cornerstone of proving correctness for recursive algorithms
  - **Base Case:** Prove the property holds for the smallest valid input
  - **Inductive Hypothesis:** Assume the property holds for input of size $k$
  - **Inductive Step:** Prove the property holds for input of size $k+1$
  - **Strong Induction:** Assume the property holds for all inputs up to size $k$

**Example – Proving Binary Search Correctness:**
```text
Theorem: Binary search returns the correct index or -1 if not found.
Base Case: $n=1$, array has one element, direct comparison works.
Inductive Step: If binary search works for arrays of size $k$, it works for arrays of size $k+1$ by comparing middle element and recursively searching the appropriate half.
```

- **Proof by Contradiction:** Assume the opposite of what we want to prove, derive a contradiction
  - **Applications:** Proving lower bounds, impossibility results, optimality
  - **Example:** Proving that comparison-based sorting requires $\Omega(n \log n)$ comparisons

- **Direct Proof:** Step-by-step logical reasoning from premises to conclusion
  - **Applications:** Proving algorithm properties, complexity bounds
  - **Example:** Proving that merge sort is stable and has $O(n \log n)$ complexity

- **Proof by Construction:** Demonstrate existence by providing an explicit example
  - **Applications:** Showing algorithms exist, proving upper bounds
  - **Example:** Constructing a sorting algorithm that achieves $O(n \log n)$ complexity

### 🔁 Recurrence Relations

Recurrence relations are the mathematical foundation for analyzing recursive algorithms. Expert practitioners must be able to solve complex recurrences to understand algorithm performance.

**Common Recurrence Patterns and Solutions:**

| Pattern                  | Recurrence                | Solution      | Example Algorithm      |
|--------------------------|--------------------------|---------------|-----------------------|
| Divide by 2              | $T(n) = T(n/2) + O(1)$   | $O(\log n)$   | Binary search         |
| Divide by 2, linear work | $T(n) = 2T(n/2) + O(n)$  | $O(n \log n)$ | Merge sort            |
| Divide by 2, const work  | $T(n) = 2T(n/2) + O(1)$  | $O(n)$        | Tree traversal        |
| Linear decrease          | $T(n) = T(n-1) + O(1)$   | $O(n)$        | Linear recursion      |
| Linear decrease, linear  | $T(n) = T(n-1) + O(n)$   | $O(n^2)$      | Selection sort        |
| Divide by 3              | $T(n) = T(n/3) + O(1)$   | $O(\log_3 n)$ | Ternary search        |

**Advanced Solution Methods:**

1. **Substitution Method:**
   - Guess the form of the solution
   - Prove by induction
   - **Example:** $T(n) = 2T(n/2) + n$, guess $T(n) = O(n \log n)$

2. **Recursion Tree Method:**
   - Visualize the recursive calls as a tree
   - Sum the costs at each level
   - **Example:** $T(n) = 3T(n/4) + n^2$

3. **Master Theorem:** Direct formula for divide-and-conquer recurrences
   - $T(n) = aT(n/b) + f(n)$ where $a \geq 1, b > 1$
   - Compare $f(n)$ with $n^{\log_b a}$
   - **Case 1:** $f(n) = O(n^{\log_b a - \varepsilon}) \implies T(n) = \Theta(n^{\log_b a})$
   - **Case 2:** $f(n) = \Theta(n^{\log_b a}) \implies T(n) = \Theta(n^{\log_b a} \log n)$
   - **Case 3:** $f(n) = \Omega(n^{\log_b a + \varepsilon}) \implies T(n) = \Theta(f(n))$

4. **Akra-Bazzi Method:** Generalization of Master Theorem for more complex recurrences

### 🎲 Probability in Algorithms

Probabilistic analysis provides crucial insights into average-case performance and enables the design of randomized algorithms. Expert practitioners use probability theory to analyze and design robust algorithms.

**Fundamental Probability Concepts:**

- **Random Variables and Expectation:**
  - **Expected Value:** $E[X] = \sum x \cdot P(X=x)$ for discrete, $E[X] = \int x \cdot f(x)dx$ for continuous
  - **Linearity of Expectation:** $E[X+Y] = E[X] + E[Y]$ (even for dependent variables)
  - **Markov's Inequality:** $P(X \geq a) \leq E[X]/a$ for non-negative $X$
  - **Chebyshev's Inequality:** $P(|X - E[X]| \geq k\sigma) \leq 1/k^2$

- **Indicator Variables:** Powerful technique for counting events
  - **Definition:** $I_A = 1$ if event $A$ occurs, $0$ otherwise
  - **Property:** $E[I_A] = P(A)$
  - **Application:** Counting expected number of inversions in random permutation

- **Concentration Inequalities:**
  - **Chernoff Bounds:** For independent random variables, $P(X \geq (1+\delta)\mu) \leq e^{-\delta^2\mu/3}$
  - **Hoeffding's Inequality:** For bounded random variables, $P(|X - E[X]| \geq t) \leq 2e^{-2t^2/n}$

- **Randomized Algorithms:**
  - **Las Vegas:** Always correct, randomized running time
  - **Monte Carlo:** Fixed running time, randomized correctness
  - **Example:** Randomized quicksort (Las Vegas), Miller-Rabin primality test (Monte Carlo)

### 🔢 Number Theory Basics

Number theory provides essential tools for cryptographic algorithms, hash functions, and advanced data structures. Expert practitioners must understand these concepts for secure and efficient algorithm design.

- **Modular Arithmetic:**
  - **Basic Operations:** $(a + b) \bmod m = ((a \bmod m) + (b \bmod m)) \bmod m$
  - **Properties:** $(a \cdot b) \bmod m = ((a \bmod m) \cdot (b \bmod m)) \bmod m$
  - **Modular Exponentiation:** $a^b \bmod m$ using fast exponentiation
  - **Modular Inverse:** $a^{-1} \bmod m$ exists if $\gcd(a,m) = 1$

- **Greatest Common Divisor (GCD):**
  - **Euclidean Algorithm:** $\gcd(a,b) = \gcd(b, a \bmod b)$
  - **Extended Euclidean Algorithm:** Find $x, y$ such that $ax + by = \gcd(a,b)$
  - **Time Complexity:** $O(\log \min(a,b))$
  - **Applications:** Modular inverse, linear Diophantine equations

- **Prime Numbers and Primality Testing:**
  - **Fundamental Theorem:** Every integer $> 1$ has unique prime factorization
  - **Sieve of Eratosthenes:** Find all primes up to $n$ in $O(n \log \log n)$
  - **Fermat's Little Theorem:** $a^{p-1} \equiv 1 \pmod p$ for prime $p$
  - **Miller-Rabin Test:** Probabilistic primality test with high accuracy

- **Chinese Remainder Theorem:**
  - **Statement:** If $\gcd(m_1, m_2) = 1$, then system $x \equiv a_1 \pmod{m_1}$, $x \equiv a_2 \pmod{m_2}$ has unique solution $\pmod{m_1 m_2}$
  - **Applications:** Large number arithmetic, cryptographic protocols

- **Discrete Logarithm:**
  - **Definition:** Find $x$ such that $g^x \equiv h \pmod p$
  - **Applications:** Diffie-Hellman key exchange, ElGamal encryption
  - **Complexity:** Believed to be hard (foundation of many cryptographic systems)

- **Advanced Number Theory:**
  - **Quadratic Residues:** Numbers that are perfect squares $\pmod p$
  - **Legendre Symbol:** $(a/p) = a^{(p-1)/2} \bmod p$
  - **Applications:** Quadratic sieve factorization, elliptic curve cryptography


````python
def analyze_time_complexity(n):
    """Example of different time complexities"""
    
    # O(1) - Constant Time
    result = n * 2
    
    # O(log n) - Logarithmic Time
    i = n
    while i > 1:
        i = i // 2
        
    # O(n) - Linear Time
    for i in range(n):
        result += i
        
    # O(n log n) - Linearithmic Time
    for i in range(n):
        j = n
        while j > 1:
            j = j // 2
            result += j
            
    # O(n²) - Quadratic Time
    for i in range(n):
        for j in range(n):
            result += i * j
            
    return result

# Example of Space Complexity Analysis
def space_complexity_examples(n):
    # O(1) Space - Constant
    sum_result = 0
    for i in range(n):
        sum_result += i
        
    # O(n) Space - Linear
    array = [i for i in range(n)]
    
    # O(n²) Space - Quadratic
    matrix = [[0 for _ in range(n)] for _ in range(n)]
    
    return sum_result, array, matrix

# Example of Amortized Analysis with Dynamic Array
class DynamicArray:
    def __init__(self):
        self.array = [None] * 2  # Initial capacity of 2
        self.size = 0
        
    def append(self, element):
        # If array is full, double its capacity
        if self.size == len(self.array):
            self._resize()
        
        self.array[self.size] = element
        self.size += 1
        
    def _resize(self):
        # Double the capacity
        new_array = [None] * (len(self.array) * 2)
        # Copy elements
        for i in range(self.size):
            new_array[i] = self.array[i]
        self.array = new_array

# Example usage
if __name__ == "__main__":
    print("Time Complexity Analysis:")
    result = analyze_time_complexity(5)
    print(f"Result: {result}")
    
    print("\nSpace Complexity Analysis:")
    sum_result, array, matrix = space_complexity_examples(3)
    print(f"Sum: {sum_result}")
    print(f"Array: {array}")
    print(f"Matrix: {matrix}")
    
    print("\nAmortized Analysis Example:")
    dynamic_array = DynamicArray()
    for i in range(5):
        dynamic_array.append(i)
        print(f"Added {i}, Current capacity: {len(dynamic_array.array)}")
````

````python
# Number Theory Examples
def gcd(a, b):
    """Greatest Common Divisor using Euclidean Algorithm"""
    while b:
        a, b = b, a % b
    return a

def extended_gcd(a, b):
    """Extended Euclidean Algorithm"""
    if a == 0:
        return b, 0, 1
    gcd, x1, y1 = extended_gcd(b % a, a)
    x = y1 - (b//a) * x1
    y = x1
    return gcd, x, y

def is_prime(n):
    """Check if a number is prime"""
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

def sieve_of_eratosthenes(n):
    """Find all prime numbers up to n"""
    sieve = [True] * (n + 1)
    sieve[0] = sieve[1] = False
    
    for i in range(2, int(n ** 0.5) + 1):
        if sieve[i]:
            for j in range(i*i, n+1, i):
                sieve[j] = False
    
    return [i for i in range(n+1) if sieve[i]]

def modular_exponentiation(base, exponent, modulus):
    """Calculate (base^exponent) % modulus efficiently"""
    result = 1
    base = base % modulus
    while exponent > 0:
        if exponent & 1:
            result = (result * base) % modulus
        base = (base * base) % modulus
        exponent >>= 1
    return result

# Probability Examples
import random

def monte_carlo_pi(num_points):
    """Estimate π using Monte Carlo method"""
    inside_circle = 0
    
    for _ in range(num_points):
        x = random.uniform(-1, 1)
        y = random.uniform(-1, 1)
        if x*x + y*y <= 1:
            inside_circle += 1
    
    pi_estimate = 4 * inside_circle / num_points
    return pi_estimate

def random_sampling(items, k):
    """Reservoir sampling: randomly select k items from a stream"""
    if k <= 0:
        return []
    
    reservoir = items[:k]
    
    for i in range(k, len(items)):
        j = random.randint(0, i)
        if j < k:
            reservoir[j] = items[i]
    
    return reservoir

def probability_simulation():
    """Simulate probability experiments"""
    # Simulate rolling two dice
    def roll_two_dice(num_rolls):
        sums = [0] * 13  # Index 0 and 1 won't be used
        for _ in range(num_rolls):
            die1 = random.randint(1, 6)
            die2 = random.randint(1, 6)
            sums[die1 + die2] += 1
        
        # Calculate probabilities
        probabilities = [count/num_rolls for count in sums[2:]]
        return probabilities
    
    return roll_two_dice(10000)

# Example usage
if __name__ == "__main__":
    # Number Theory Examples
    print("Number Theory Examples:")
    a, b = 48, 36
    print(f"GCD of {a} and {b}: {gcd(a, b)}")
    
    gcd_val, x, y = extended_gcd(a, b)
    print(f"Extended GCD of {a} and {b}: {gcd_val}")
    print(f"Coefficients x, y: {x}, {y}")
    
    n = 17
    print(f"Is {n} prime? {is_prime(n)}")
    
    n = 30
    primes = sieve_of_eratosthenes(n)
    print(f"Prime numbers up to {n}: {primes}")
    
    base, exp, mod = 2, 10, 1000
    result = modular_exponentiation(base, exp, mod)
    print(f"{base}^{exp} % {mod} = {result}")
    
    # Probability Examples
    print("\nProbability Examples:")
    num_points = 100000
    pi_estimate = monte_carlo_pi(num_points)
    print(f"π estimate using {num_points} points: {pi_estimate}")
    
    items = list(range(1, 11))
    k = 4
    sample = random_sampling(items, k)
    print(f"Random sample of {k} items from {items}: {sample}")
    
    dice_probabilities = probability_simulation()
    print("Probabilities for sums of two dice:")
    for i, prob in enumerate(dice_probabilities):
        print(f"Sum {i+2}: {prob:.3f}")
````

````python
# Binary Tree Implementation
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class BinaryTree:
    def __init__(self):
        self.root = None
    
    def traversal_iterative(self, method='inorder'):
        """Iterative tree traversal"""
        if not self.root:
            return []
        
        result = []
        stack = []
        current = self.root
        
        if method == 'inorder':
            while current or stack:
                while current:
                    stack.append(current)
                    current = current.left
                current = stack.pop()
                result.append(current.val)
                current = current.right
                
        elif method == 'preorder':
            stack.append(self.root)
            while stack:
                current = stack.pop()
                result.append(current.val)
                if current.right:
                    stack.append(current.right)
                if current.left:
                    stack.append(current.left)
                    
        elif method == 'postorder':
            stack.append(self.root)
            while stack:
                current = stack.pop()
                result.insert(0, current.val)
                if current.left:
                    stack.append(current.left)
                if current.right:
                    stack.append(current.right)
        
        return result

# AVL Tree Implementation
class AVLNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None
        self.height = 1

class AVLTree:
    def get_height(self, node):
        if not node:
            return 0
        return node.height
    
    def get_balance(self, node):
        if not node:
            return 0
        return self.get_height(node.left) - self.get_height(node.right)
    
    def right_rotate(self, y):
        x = y.left
        T2 = x.right
        
        x.right = y
        y.left = T2
        
        y.height = max(self.get_height(y.left), 
                      self.get_height(y.right)) + 1
        x.height = max(self.get_height(x.left), 
                      self.get_height(x.right)) + 1
        
        return x
    
    def left_rotate(self, x):
        y = x.right
        T2 = y.left
        
        y.left = x
        x.right = T2
        
        x.height = max(self.get_height(x.left), 
                      self.get_height(x.right)) + 1
        y.height = max(self.get_height(y.left), 
                      self.get_height(y.right)) + 1
        
        return y
    
    def insert(self, root, val):
        # Standard BST insert
        if not root:
            return AVLNode(val)
            
        if val < root.val:
            root.left = self.insert(root.left, val)
        elif val > root.val:
            root.right = self.insert(root.right, val)
        else:
            return root  # Duplicate values not allowed
        
        # Update height
        root.height = max(self.get_height(root.left),
                         self.get_height(root.right)) + 1
        
        # Get balance factor
        balance = self.get_balance(root)
        
        # Left Left Case
        if balance > 1 and val < root.left.val:
            return self.right_rotate(root)
        
        # Right Right Case
        if balance < -1 and val > root.right.val:
            return self.left_rotate(root)
        
        # Left Right Case
        if balance > 1 and val > root.left.val:
            root.left = self.left_rotate(root.left)
            return self.right_rotate(root)
        
        # Right Left Case
        if balance < -1 and val < root.right.val:
            root.right = self.right_rotate(root.right)
            return self.left_rotate(root)
        
        return root

# Red-Black Tree Implementation
class Color:
    RED = True
    BLACK = False

class RBNode:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None
        self.parent = None
        self.color = Color.RED

class RedBlackTree:
    def __init__(self):
        self.NIL = RBNode(None)
        self.NIL.color = Color.BLACK
        self.root = self.NIL
    
    def left_rotate(self, x):
        y = x.right
        x.right = y.left
        if y.left != self.NIL:
            y.left.parent = x
        y.parent = x.parent
        if x.parent == self.NIL:
            self.root = y
        elif x == x.parent.left:
            x.parent.left = y
        else:
            x.parent.right = y
        y.left = x
        x.parent = y
    
    def right_rotate(self, x):
        y = x.left
        x.left = y.right
        if y.right != self.NIL:
            y.right.parent = x
        y.parent = x.parent
        if x.parent == self.NIL:
            self.root = y
        elif x == x.parent.right:
            x.parent.right = y
        else:
            x.parent.left = y
        y.right = x
        x.parent = y
    
    def insert(self, val):
        node = RBNode(val)
        node.left = self.NIL
        node.right = self.NIL
        
        y = self.NIL
        x = self.root
        
        while x != self.NIL:
            y = x
            if node.val < x.val:
                x = x.left
            else:
                x = x.right
        
        node.parent = y
        if y == self.NIL:
            self.root = node
        elif node.val < y.val:
            y.left = node
        else:
            y.right = node
        
        self.insert_fixup(node)
    
    def insert_fixup(self, node):
        while node.parent.color == Color.RED:
            if node.parent == node.parent.parent.left:
                y = node.parent.parent.right
                if y.color == Color.RED:
                    node.parent.color = Color.BLACK
                    y.color = Color.BLACK
                    node.parent.parent.color = Color.RED
                    node = node.parent.parent
                else:
                    if node == node.parent.right:
                        node = node.parent
                        self.left_rotate(node)
                    node.parent.color = Color.BLACK
                    node.parent.parent.color = Color.RED
                    self.right_rotate(node.parent.parent)
            else:
                y = node.parent.parent.left
                if y.color == Color.RED:
                    node.parent.color = Color.BLACK
                    y.color = Color.BLACK
                    node.parent.parent.color = Color.RED
                    node = node.parent.parent
                else:
                    if node == node.parent.left:
                        node = node.parent
                        self.right_rotate(node)
                    node.parent.color = Color.BLACK
                    node.parent.parent.color = Color.RED
                    self.left_rotate(node.parent.parent)
        
        self.root.color = Color.BLACK

# B-Tree Implementation
class BTreeNode:
    def __init__(self, leaf=True):
        self.leaf = leaf
        self.keys = []
        self.children = []

class BTree:
    def __init__(self, t):
        self.root = BTreeNode()
        self.t = t  # Minimum degree
    
    def insert(self, k):
        root = self.root
        if len(root.keys) == (2 * self.t) - 1:
            new_root = BTreeNode(leaf=False)
            new_root.children.append(self.root)
            self.split_child(new_root, 0)
            self.root = new_root
        self.insert_non_full(self.root, k)
    
    def split_child(self, x, i):
        t = self.t
        y = x.children[i]
        z = BTreeNode(leaf=y.leaf)
        
        x.keys.insert(i, y.keys[t-1])
        x.children.insert(i+1, z)
        
        z.keys = y.keys[t:]
        y.keys = y.keys[:t-1]
        
        if not y.leaf:
            z.children = y.children[t:]
            y.children = y.children[:t]
    
    def insert_non_full(self, x, k):
        i = len(x.keys) - 1
        if x.leaf:
            while i >= 0 and k < x.keys[i]:
                i -= 1
            x.keys.insert(i+1, k)
        else:
            while i >= 0 and k < x.keys[i]:
                i -= 1
            i += 1
            if len(x.children[i].keys) == (2 * self.t) - 1:
                self.split_child(x, i)
                if k > x.keys[i]:
                    i += 1
            self.insert_non_full(x.children[i], k)

# Advanced Data Structures

# Min Heap Implementation
class MinHeap:
    def __init__(self):
        self.heap = []
    
    def parent(self, i):
        return (i - 1) // 2
    
    def left_child(self, i):
        return 2 * i + 1
    
    def right_child(self, i):
        return 2 * i + 2
    
    def swap(self, i, j):
        self.heap[i], self.heap[j] = self.heap[j], self.heap[i]
    
    def insert(self, key):
        self.heap.append(key)
        self._sift_up(len(self.heap) - 1)
    
    def extract_min(self):
        if not self.heap:
            return None
        
        min_val = self.heap[0]
        last_val = self.heap.pop()
        
        if self.heap:
            self.heap[0] = last_val
            self._sift_down(0)
        
        return min_val
    
    def _sift_up(self, i):
        parent = self.parent(i)
        if i > 0 and self.heap[i] < self.heap[parent]:
            self.swap(i, parent)
            self._sift_up(parent)
    
    def _sift_down(self, i):
        min_idx = i
        left = self.left_child(i)
        right = self.right_child(i)
        
        if (left < len(self.heap) and 
            self.heap[left] < self.heap[min_idx]):
            min_idx = left
            
        if (right < len(self.heap) and 
            self.heap[right] < self.heap[min_idx]):
            min_idx = right
            
        if min_idx != i:
            self.swap(i, min_idx)
            self._sift_down(min_idx)

# Segment Tree Implementation
class SegmentTree:
    def __init__(self, arr):
        self.n = len(arr)
        self.tree = [0] * (4 * self.n)
        self._build_tree(arr, 0, 0, self.n - 1)
    
    def _build_tree(self, arr, node, start, end):
        if start == end:
            self.tree[node] = arr[start]
            return
        
        mid = (start + end) // 2
        self._build_tree(arr, 2*node + 1, start, mid)
        self._build_tree(arr, 2*node + 2, mid + 1, end)
        self.tree[node] = self.tree[2*node + 1] + self.tree[2*node + 2]
    
    def update(self, index, value):
        self._update_tree(0, 0, self.n - 1, index, value)
    
    def _update_tree(self, node, start, end, idx, val):
        if start == end:
            self.tree[node] = val
            return
        
        mid = (start + end) // 2
        if start <= idx <= mid:
            self._update_tree(2*node + 1, start, mid, idx, val)
        else:
            self._update_tree(2*node + 2, mid + 1, end, idx, val)
        
        self.tree[node] = self.tree[2*node + 1] + self.tree[2*node + 2]
    
    def query(self, left, right):
        return self._query_tree(0, 0, self.n - 1, left, right)
    
    def _query_tree(self, node, start, end, l, r):
        if r < start or end < l:
            return 0
        
        if l <= start and end <= r:
            return self.tree[node]
        
        mid = (start + end) // 2
        left_sum = self._query_tree(2*node + 1, start, mid, l, r)
        right_sum = self._query_tree(2*node + 2, mid + 1, end, l, r)
        return left_sum + right_sum

# Trie Implementation
class TrieNode:
    def __init__(self):
        self.children = {}
        self.is_end = False

class Trie:
    def __init__(self):
        self.root = TrieNode()
    
    def insert(self, word):
        node = self.root
        for char in word:
            if char not in node.children:
                node.children[char] = TrieNode()
            node = node.children[char]
        node.is_end = True
    
    def search(self, word):
        node = self.root
        for char in word:
            if char not in node.children:
                return False
            node = node.children[char]
        return node.is_end
    
    def starts_with(self, prefix):
        node = self.root
        for char in prefix:
            if char not in node.children:
                return False
            node = node.children[char]
        return True

# Example usage
if __name__ == "__main__":
    # Binary Tree Example
    print("Binary Tree Example:")
    bt = BinaryTree()
    bt.root = TreeNode(1)
    bt.root.left = TreeNode(2)
    bt.root.right = TreeNode(3)
    bt.root.left.left = TreeNode(4)
    bt.root.left.right = TreeNode(5)
    
    print("Inorder traversal:", bt.traversal_iterative('inorder'))
    print("Preorder traversal:", bt.traversal_iterative('preorder'))
    print("Postorder traversal:", bt.traversal_iterative('postorder'))
    
    # AVL Tree Example
    print("\nAVL Tree Example:")
    avl = AVLTree()
    root = None
    values = [10, 20, 30, 40, 50, 25]
    
    for val in values:
        root = avl.insert(root, val)
    print("AVL Tree root value:", root.val)
    print("AVL Tree root height:", root.height)
    
    # Red-Black Tree Example
    print("\nRed-Black Tree Example:")
    rb = RedBlackTree()
    values = [7, 3, 18, 10, 22, 8, 11, 26, 2, 6, 13]
    
    for val in values:
        rb.insert(val)
    print("Red-Black Tree root value:", rb.root.val)
    print("Root color:", "RED" if rb.root.color else "BLACK")
    
    # B-Tree Example
    print("\nB-Tree Example:")
    b = BTree(3)  # Minimum degree = 3
    values = [10, 20, 5, 6, 12, 30, 7, 17]
    
    for val in values:
        b.insert(val)
    print("B-Tree root keys:", b.root.keys)
    
    # Min Heap Example
    print("\nMin Heap Example:")
    heap = MinHeap()
    values = [3, 2, 1, 15, 5, 4, 45]
    
    for val in values:
        heap.insert(val)
    
    print("Extract min operations:")
    for _ in range(3):
        print(heap.extract_min())
    
    # Segment Tree Example
    print("\nSegment Tree Example:")
    arr = [1, 3, 5, 7, 9, 11]
    st = SegmentTree(arr)
    
    print("Sum of values in range [1, 3]:", st.query(1, 3))
    st.update(1, 10)
    print("Sum after updating index 1 to 10:", st.query(1, 3))
    
    # Trie Example
    print("\nTrie Example:")
    trie = Trie()
    words = ["apple", "app", "apricot", "banana"]
    
    for word in words:
        trie.insert(word)
    
    print("Search 'apple':", trie.search("apple"))
    print("Search 'app':", trie.search("app"))
    print("Prefix 'apr':", trie.starts_with("apr"))
````

# 🧠 III. Algorithm Paradigms

> 💡 **Algorithm paradigms** are fundamental approaches to solving computational problems. Each paradigm provides a systematic way to think about and structure solutions, making them essential tools for expert-level problem-solving.

## 🔄 1. Divide and Conquer

**Divide and Conquer** is a fundamental algorithmic paradigm that breaks down problems into smaller subproblems, solves them recursively, and combines their solutions.

### 🎯 Core Principles

| Principle         | Description                                   | Example                   |
|-------------------|-----------------------------------------------|---------------------------|
| 🔀 **Divide**     | Break the problem into smaller, similar subproblems | Split array into two halves |
| ⚔️ **Conquer**    | Solve the subproblems recursively             | Sort each half independently |
| 🔗 **Combine**    | Merge the solutions of subproblems to solve the original problem | Merge two sorted halves |

### 📊 Time Complexity Analysis

For a problem of size $n$ divided into $a$ subproblems of size $n/b$:

- 🧮 **Recurrence Relation**: $T(n) = aT(n/b) + f(n)$
- 📐 **Master Theorem**: $T(n) = O(n^{\log_b a})$ if $f(n) = O(n^{\log_b a - \varepsilon})$

> 💡 **Key Insight**: The efficiency depends on how well we can divide the problem and how expensive the combine step is.

### 🕵️‍♂️ Binary Search Variations

Binary search is a fundamental divide-and-conquer technique that can be adapted to solve various problems beyond simple sorted array searches.

**Core Binary Search Template:**
```python
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
```

**Advanced Variations:**

1. **🎯 Finding First/Last Occurrence**
   - Useful for range queries
   - Handles duplicates efficiently
   - Time complexity: $O(\log n)$

   - Array rotated at unknown pivot
   - Modified binary search condition
   - Time complexity: $O(\log n)$

3. **🔄 Peak Element Finding**
   - Find local maximum in array
   - Uses neighbor comparison
   - Time complexity: $O(\log n)$

4. **📈 Binary Search on Answer**
   - Solve optimization problems
   - Search space is answer range
   - Time complexity: $O(\log range \times validation)$

## 🎯 2. Dynamic Programming

**Dynamic Programming** is a method for solving complex problems by breaking them down into simpler subproblems and storing the results to avoid redundant calculations.

> 🧠 **Core Philosophy**: "Those who cannot remember the past are condemned to repeat it" - Store solutions to avoid recomputation!

### 🏗️ Core Principles

| Principle | Description | Example |
|-----------|-------------|---------|
| **🔗 Optimal Substructure** | Optimal solution contains optimal solutions to subproblems | Fibonacci: F(n) = F(n-1) + F(n-2) |
| **🔄 Overlapping Subproblems** | Same subproblems are solved multiple times | Fibonacci: F(3) calculated multiple times |
| **💾 Memoization/Tabulation** | Store solutions to avoid recomputation | Cache results in table or dictionary |

### 🎨 Implementation Approaches

1. **📝 Memoization (Top-Down)**
   - Recursive approach with caching
   - Natural problem-solving flow
   - Cache only needed states
   - Stack space overhead

2. **📊 Tabulation (Bottom-Up)**
   - Iterative approach
   - Fill DP table systematically
   - More space-efficient
   - Often faster in practice

3. **⚡ Space Optimization**
   - Rolling array technique
   - State compression
   - Memory-efficient variants
   - Trade space for time

### 🎯 Advanced DP Techniques

1. **State Design**
   - 🎨 Minimize state dimensions
   - 🔄 Ensure state uniqueness
   - 💡 Handle transitions efficiently
   - 📊 Consider space-time tradeoffs

2. **Optimization Methods**
   - 🎯 State compression
   - 💾 Memory-efficient solutions
   - 🚀 Cache-friendly access patterns
   - 📈 Parallel DP when possible

3. **Common Patterns**
   - 🔄 Interval DP
   - 🔢 Digit DP
   - 🌳 Tree DP
   - 🎭 Bitmask DP

> 💡 **Best Practice**: Always start with a clear state definition and transition function before implementation.

### 🎨 State Design and Transitions

Understanding how to design states and transitions is crucial for effective dynamic programming solutions.

> 🔑 **Key Question**: "What information do I need to uniquely identify a subproblem?"

#### State Design Principles:
- **🎯 Minimal State**: Include only necessary information
- **🔄 Clear Transitions**: Define how to move between states
- **📊 Base Cases**: Identify the simplest subproblems
- **⚡ Optimization**: Consider space-time trade-offs

#### Common State Patterns:
- **1D States**: Single parameter (e.g., position, index)
- **2D States**: Two parameters (e.g., position + capacity)
- **Multi-dimensional**: Complex state spaces

## 🎯 3. Greedy Algorithms

**Greedy algorithms** make locally optimal choices at each step with the hope of finding a global optimum.

> 🍯 **Greedy Philosophy**: "Take the best choice now, and trust that it leads to the best overall solution!"

### 🏗️ Core Principles

| Principle | Description | Example |
|-----------|-------------|---------|
| **🎯 Greedy Choice Property** | A global optimum can be arrived at by making locally optimal choices | Always pick the activity that ends earliest |
| **🔗 Optimal Substructure** | Optimal solution contains optimal solutions to subproblems | If we pick the best activity now, the remaining problem is still optimal |
| **🚫 No Backtracking** | Once a choice is made, it's never reconsidered | We never go back to change previous decisions |

### ⚡ When to Use Greedy

- **✅ Clear Greedy Choice**: There's an obvious "best" choice at each step
- **✅ Optimal Substructure**: The problem breaks down into similar subproblems
- **✅ No Overlapping Subproblems**: Unlike DP, we don't need to store solutions
- **❌ Avoid When**: Local optimum doesn't guarantee global optimum

### 📦 Huffman Coding

Huffman coding is a classic greedy algorithm for data compression that creates optimal prefix codes.

> 🗜️ **Goal**: Create the most efficient binary encoding for characters based on their frequency.

#### Algorithm Steps:
1. **📊 Count frequencies** of each character
2. **🌳 Build min-heap** with character nodes
3. **🔄 While heap has > 1 node**:
   - Extract two nodes with minimum frequency
   - Create new internal node with combined frequency
   - Insert back into heap
4. **📝 Generate codes** by traversing the tree

#### Greedy Choice:
- **Always merge the two least frequent characters first**
- This minimizes the total encoding length

> 💡 **Why it's optimal**: Characters with higher frequency get shorter codes, reducing overall message length!

#### Complexity:
- **Time**: O(n log n) where n is the number of unique characters
- **Space**: O(n) for the tree and codes

### 🌳 Minimum Spanning Trees

Greedy algorithms are perfect for finding minimum spanning trees in graphs.

> 🎯 **Goal**: Find the minimum weight tree that connects all vertices in a weighted graph.

#### Two Classic Greedy Approaches:

| Algorithm | Greedy Choice | Time Complexity | Best For |
|-----------|---------------|-----------------|----------|
| **🔗 Kruskal's** | Always add the minimum weight edge that doesn't create a cycle | O(E log E) | Sparse graphs |
| **🌱 Prim's** | Always add the minimum weight edge from current MST to new vertex | O(E log V) | Dense graphs |

#### Kruskal's Algorithm:
1. **📊 Sort edges** by weight (ascending)
2. **🔄 For each edge** (in order):
   - If it doesn't create a cycle, add it to MST
   - Use Union-Find to detect cycles efficiently

> 💡 **Greedy Insight**: The minimum weight edge that doesn't create a cycle must be part of the MST!

#### Applications:
- **🌐 Network design** (minimum cost to connect all nodes)
- **🗺️ Circuit design** (minimum wire length)
- **📡 Clustering** (group similar data points)


### ⏰ Interval Scheduling

Interval scheduling problems are classic examples where greedy algorithms provide optimal solutions.

> 🎯 **Two Main Variants**:
> 1. **Maximum Non-overlapping Intervals**: Select maximum number of non-conflicting intervals
> 2. **Minimum Resources**: Find minimum number of resources needed for all intervals

#### Variant 1: Maximum Non-overlapping Intervals
- **Greedy Strategy**: Sort by finish time, always pick the earliest finishing interval
- **Why it works**: Picking the earliest finish leaves maximum room for future intervals

#### Variant 2: Minimum Resources (Interval Partitioning)
- **Greedy Strategy**: Sort by start time, assign each interval to the earliest available resource
- **Why it works**: This minimizes the number of resources needed

> 💡 **Key Insight**: The greedy choice is always the "most constrained" option that still works!

#### Complexity:
- **Time**: O(n log n) for sorting
- **Space**: O(1) for maximum intervals, O(k) for minimum resources where k is the answer

````python
# Divide and Conquer Examples

def binary_search(arr, target):
    """Binary Search Implementation"""
    def binary_search_helper(left, right):
        if left > right:
            return -1
        
        mid = left + (right - left) // 2
        
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            return binary_search_helper(mid + 1, right)
        else:
            return binary_search_helper(left, mid - 1)
    
    return binary_search_helper(0, len(arr) - 1)

def merge_sort(arr):
    """Merge Sort Implementation"""
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    return merge(left, right)

def merge(left, right):
    """Helper function for merge sort"""
    result = []
    i = j = 0
    
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    result.extend(left[i:])
    result.extend(right[j:])
    return result

# Dynamic Programming Examples

def fibonacci_dp(n):
    """Fibonacci with Dynamic Programming"""
    if n <= 1:
        return n
    
    dp = [0] * (n + 1)
    dp[1] = 1
    
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]
    
    return dp[n]

def knapsack(values, weights, capacity):
    """0/1 Knapsack Problem"""
    n = len(values)
    dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]
    
    for i in range(1, n + 1):
        for w in range(capacity + 1):
            if weights[i-1] <= w:
                dp[i][w] = max(values[i-1] + dp[i-1][w-weights[i-1]], 
                             dp[i-1][w])
            else:
                dp[i][w] = dp[i-1][w]
    
    return dp[n][capacity]

def longest_common_subsequence(text1, text2):
    """Longest Common Subsequence"""
    m, n = len(text1), len(text2)
    dp = [[0] * (n + 1) for _ in range(m + 1)]
    
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if text1[i-1] == text2[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])
    
    return dp[m][n]

# Greedy Algorithm Examples

def activity_selection(start, finish):
    """Activity Selection Problem"""
    n = len(start)
    activities = list(zip(start, finish))
    # Sort by finish time
    activities.sort(key=lambda x: x[1])
    
    selected = [activities[0]]
    last_finish = activities[0][1]
    
    for i in range(1, n):
        if activities[i][0] >= last_finish:
            selected.append(activities[i])
            last_finish = activities[i][1]
    
    return selected

def fractional_knapsack(values, weights, capacity):
    """Fractional Knapsack Problem"""
    n = len(values)
    # Calculate value per unit weight
    ratios = [(values[i]/weights[i], weights[i], values[i]) 
              for i in range(n)]
    ratios.sort(reverse=True)  # Sort by value/weight ratio
    
    total_value = 0
    remaining_capacity = capacity
    
    for ratio, weight, value in ratios:
        if remaining_capacity >= weight:
            # Take the whole item
            total_value += value
            remaining_capacity -= weight
        else:
            # Take fraction of the item
            total_value += ratio * remaining_capacity
            break
    
    return total_value

# Example usage
if __name__ == "__main__":
    # Divide and Conquer Examples
    print("Divide and Conquer Examples:")
    arr = [1, 3, 5, 7, 9, 11, 13, 15]
    target = 7
    print(f"Binary Search: Found {target} at index {binary_search(arr, target)}")
    
    unsorted = [64, 34, 25, 12, 22, 11, 90]
    print(f"Merge Sort: {merge_sort(unsorted)}")
    
    # Dynamic Programming Examples
    print("\nDynamic Programming Examples:")
    n = 10
    print(f"Fibonacci({n}) = {fibonacci_dp(n)}")
    
    values = [60, 100, 120]
    weights = [10, 20, 30]
    capacity = 50
    print(f"Knapsack max value: {knapsack(values, weights, capacity)}")
    
    text1, text2 = "ABCDGH", "AEDFHR"
    print(f"LCS length: {longest_common_subsequence(text1, text2)}")
    
    # Greedy Algorithm Examples
    print("\nGreedy Algorithm Examples:")
    start = [1, 3, 0, 5, 8, 5]
    finish = [2, 4, 6, 7, 9, 9]
    selected = activity_selection(start, finish)
    print(f"Selected activities: {selected}")
    
    values = [60, 100, 120]
    weights = [10, 20, 30]
    capacity = 50
    max_value = fractional_knapsack(values, weights, capacity)
    print(f"Fractional Knapsack max value: {max_value}")
````

#### 🪟 Sliding Window
The sliding window technique maintains a window of elements and efficiently updates it as we traverse the data structure.

**When to Use**:
- Subarray/substring problems with size constraints
- Finding maximum/minimum subarray of given size
- Problems involving consecutive elements
- String problems with character frequency

**Types of Sliding Windows**:

| Type | Description | Window Size | Use Case |
|------|-------------|-------------|----------|
| **Fixed Size** | Window size remains constant | k | Maximum sum of subarray of size k |
| **Variable Size** | Window size changes based on condition | Dynamic | Longest substring with k distinct characters |
| **Growing Window** | Window only grows | 0 to n | Smallest subarray with sum ≥ target |
| **Shrinking Window** | Window only shrinks | n to 0 | Longest valid substring |

**Implementation Template**:
```python
def sliding_window_pattern(arr, k):
    window_start = 0
    window_sum = 0
    max_sum = float('-inf')
    
    for window_end in range(len(arr)):
        window_sum += arr[window_end]
        
        # Shrink window if needed
        while window_end - window_start + 1 > k:
            window_sum -= arr[window_start]
            window_start += 1
        
        # Update result
        if window_end - window_start + 1 == k:
            max_sum = max(max_sum, window_sum)
    
    return max_sum
```

#### 🇳🇱 Dutch National Flag
The Dutch National Flag algorithm efficiently partitions an array into three sections.

**When to Use**:

```python
def dfs_iterative(graph, start):
    stack = [start]
    visited = set()
    
    while stack:
        node = stack.pop()
        if node not in visited:
            visited.add(node)
            for neighbor in graph[node]:
                if neighbor not in visited:
                    stack.append(neighbor)
```

#### 🧭 BFS Applications
Breadth-First Search guarantees finding the shortest path in unweighted graphs.

**Common Applications**:

- **Minimum Steps**: Find minimum steps to reach target

```python
def bfs(graph, start):
    queue = deque([start])
    visited = set([start])
    distance = {start: 0}
    
    while queue:
        node = queue.popleft()
        
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                distance[neighbor] = distance[node] + 1
                queue.append(neighbor)
    
    return distance
```

````python
# Array and String Patterns

def two_sum(nums, target):
    """Two Pointers Example: Find two numbers that sum to target"""
    left, right = 0, len(nums) - 1
    while left < right:
        current_sum = nums[left] + nums[right]
        if current_sum == target:
            return [left, right]
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return []

def max_sliding_window(nums, k):
    """Sliding Window Example: Find maximum in each window of size k"""
    from collections import deque
    result = []
    window = deque()  # Store indices
    
    for i in range(len(nums)):
        # Remove indices outside current window
        while window and window[0] < i - k + 1:
            window.popleft()
            
        # Remove smaller elements from back
        while window and nums[window[-1]] < nums[i]:
            window.pop()
            
        window.append(i)
        
        if i >= k - 1:
            result.append(nums[window[0]])
            
    return result

def prefix_sum_array(arr):
    """Prefix Sum Example: Calculate cumulative sum array"""
    prefix = [0] * (len(arr) + 1)
    for i in range(len(arr)):
        prefix[i + 1] = prefix[i] + arr[i]
    return prefix

def dutch_flag_partition(arr):
    """Dutch National Flag: Sort array of 0s, 1s, and 2s"""
    low = mid = 0
    high = len(arr) - 1
    
    while mid <= high:
        if arr[mid] == 0:
            arr[low], arr[mid] = arr[mid], arr[low]
            low += 1
            mid += 1
        elif arr[mid] == 1:
            mid += 1
        else:  # arr[mid] == 2
            arr[mid], arr[high] = arr[high], arr[mid]
            high -= 1
    
    return arr

# Graph Algorithms

def dfs(graph, start, visited=None):
    """Depth-First Search Implementation"""
    if visited is None:
        visited = set()
    
    visited.add(start)
    print(start, end=' ')
    
    for next_vertex in graph[start]:
        if next_vertex not in visited:
            dfs(graph, next_vertex, visited)
    
    return visited

def bfs(graph, start):
    """Breadth-First Search Implementation"""
    from collections import deque
    
    visited = {start}
    queue = deque([start])
    
    while queue:
        vertex = queue.popleft()
        print(vertex, end=' ')
        
        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    
    return visited

def dijkstra(graph, start):
    """Dijkstra's Shortest Path Algorithm"""
    import heapq
    
    distances = {vertex: float('infinity') for vertex in graph}
    distances[start] = 0
    pq = [(0, start)]
    
    while pq:
        current_distance, current_vertex = heapq.heappop(pq)
        
        if current_distance > distances[current_vertex]:
            continue
            
        for neighbor, weight in graph[current_vertex].items():
            distance = current_distance + weight
            
            if distance < distances[neighbor]:
                distances[neighbor] = distance
                heapq.heappush(pq, (distance, neighbor))
    
    return distances

# Advanced Patterns

def count_set_bits(n):
    """Bit Manipulation Example: Count number of 1s in binary"""
    count = 0
    while n:
        count += n & 1
        n >>= 1
    return count

def is_power_of_two(n):
    """Bit Manipulation Example: Check if number is power of 2"""
    return n > 0 and (n & (n - 1)) == 0

def rabin_karp(text, pattern):
    """String Matching Example: Rabin-Karp Algorithm"""
    if not pattern or not text:
        return []
    
    # Prime number for hash calculation
    prime = 101
    # Pattern length
    m = len(pattern)
    # Text length
    n = len(text)
    
    # Calculate pattern hash
    pattern_hash = 0
    for i in range(m):
        pattern_hash = (pattern_hash + ord(pattern[i])) % prime
    
    # Calculate initial window hash
    window_hash = 0
    for i in range(m):
        window_hash = (window_hash + ord(text[i])) % prime
    
    result = []
    # Check each window
    for i in range(n - m + 1):
        if pattern_hash == window_hash:
            # Verify character by character
            if text[i:i+m] == pattern:
                result.append(i)
        
        if i < n - m:
            # Calculate next window hash
            window_hash = (window_hash - ord(text[i]) + ord(text[i + m])) % prime
    
    return result

# Example usage
if __name__ == "__main__":
    # Two Pointers Example
    print("Two Pointers Example:")
    nums = [2, 7, 11, 15]
    target = 9
    result = two_sum(nums, target)
    print(f"Two numbers that sum to {target}: indices {result}")
    
    # Sliding Window Example
    print("\nSliding Window Example:")
    nums = [1, 3, -1, -3, 5, 3, 6, 7]
    k = 3
    result = max_sliding_window(nums, k)
    print(f"Maximum in each window of size {k}: {result}")
    
    # Prefix Sum Example
    print("\nPrefix Sum Example:")
    arr = [1, 2, 3, 4, 5]
    prefix = prefix_sum_array(arr)
    print(f"Prefix sum array: {prefix}")
    
    # Dutch National Flag Example
    print("\nDutch National Flag Example:")
    arr = [2, 0, 1, 2, 1, 0]
    result = dutch_flag_partition(arr.copy())
    print(f"Sorted array: {result}")
    
    # Graph Algorithms Example
    print("\nGraph Algorithms Example:")
    graph = {
        'A': ['B', 'C'],
        'B': ['A', 'D', 'E'],
        'C': ['A', 'F'],
        'D': ['B'],
        'E': ['B', 'F'],
        'F': ['C', 'E']
    }
    
    print("DFS starting from 'A':", end=' ')
    dfs(graph, 'A')
    print("\nBFS starting from 'A':", end=' ')
    bfs(graph, 'A')
    
    # Weighted Graph for Dijkstra's
    weighted_graph = {
        'A': {'B': 4, 'C': 2},
        'B': {'A': 4, 'C': 1, 'D': 5},
        'C': {'A': 2, 'B': 1, 'D': 8, 'E': 10},
        'D': {'B': 5, 'C': 8, 'E': 2},
        'E': {'C': 10, 'D': 2}
    }
    shortest_paths = dijkstra(weighted_graph, 'A')
    print(f"\nShortest paths from 'A': {shortest_paths}")
    
    # Bit Manipulation Examples
    print("\nBit Manipulation Examples:")
    n = 7  # Binary: 111
    print(f"Number of set bits in {n}: {count_set_bits(n)}")
    print(f"Is {n} a power of 2? {is_power_of_two(n)}")
    
    # String Matching Example
    print("\nString Matching Example:")
    text = "AABAACAADAABAAABAA"
    pattern = "AABA"
    matches = rabin_karp(text, pattern)
    print(f"Pattern '{pattern}' found at indices: {matches}")
````
